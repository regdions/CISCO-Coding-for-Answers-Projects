# What is the First Day of the Week?

Whether the Gregorian calendar shows Sunday or Monday as the first day of the week depends on where you live.

Do more countries start the week on Sunday or Monday? What about people? What about by continent?

The file `first-day-of-week.csv` shows the first day of the week for each territory. The file `population.csv` shows the population in the year 2020 for each territory in millions, and the file `four-regions.csv` specifies whether each territory is in asia, europe, africa, or the americas.

```python
import pandas as pd
df = pd.read_csv('first-day-of-week.csv')
df.head()
```

```python
pop = pd.read_csv('population.csv')
pop.head()
```

```python
regions = pd.read_csv('four-regions.csv')
regions.head()
```

### Project Ideas

- How many territories show Friday, Saturday, Sunday, and Monday as the `first_day` of the week?

- How many people start the week on Friday, Saturday, Sunday, and Monday?

- Which of the `four_regions` predominantly start the week on Sunday? On Monday? Are there any regions that are more divided between Sunday and Monday?

**QUESTION 1: How many territories show Friday, Saturday, Sunday, and Monday as the `first_day` of the week?**

```python
df.groupby('first_day').size()
```

**ANSWER:** The results show that Monday is the most as the first day of the week, with a rcord of 186 terrorities using it. Followed by Sunday, with 55 territories, Saturday with 15 territories, and lastly by Friday with only one sole territory using it as first day of the week.

**QUESTION 2:  How many people start the week on Friday, Saturday, Sunday, and Monday?**

```python
merged = df.merge(pop, on = 'alpha3', how = 'left')
pop_by_fday = merged.groupby('first_day')['population'].sum()
pop_by_fday
```

**ANSWER:** Majority of the population use Sunday as the first day of the week, recording 3844.08 million people. It is then followed by Monday with 3600.19 million people, Saturday with 431.79 million people. Friday is used by the least portion of the population, recording 0.50 million people.

**QUESTION 3: Which of the `four_regions` predominantly start the week on Sunday? On Monday? Are there any regions that are more divided between Sunday and Monday?**

**3.1 Region that predominantly start the week on Sunday:**

```python
region = df.merge(regions, on = 'alpha3', how = 'left')
sun = region.query('first_day == "sun"').groupby('four_regions').size().reset_index(name='sun_count')
sun
```

**ANSWER:** Asia seem to predominantly start the week on Sunday since it has the most territories that use Sunday compared to other regions.

**3.2 Region that predominantly start the week on Monday:**

```python
mon = region.query('first_day == "mon"').groupby('four_regions').size().reset_index(name='mon_count')
mon
```

**ANSWER:** Europe use Monday predominantly start their week on Monday compared to the other regions, with a total of 46 territories using it.

**3.3 Are there any regions that are more divided between Sunday and Monday?**

```python
# merging the two dataframes
sunmon = sun.merge(mon, on = 'four_regions', how = 'left')
sunmon
```

```python
#solving for the ratio of sun and mon first day of week for each region
sunmon['ratio'] = (sunmon['sun_count']/sunmon['mon_count'])

#selecting which ratio is closer to 1 by subtravting the ratio from 1 and taking the absolute value
#the value closer to zero is the one that is closer to 1
sunmon['closer_ratio'] = abs(sunmon['ratio']-1)
sunmon
```

**ANSWER:** Asia is the most divided region between Sunday and Monday as the first day of the week compared to other regions.
