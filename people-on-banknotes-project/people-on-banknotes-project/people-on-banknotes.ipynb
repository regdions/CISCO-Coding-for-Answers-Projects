# People on Banknotes

Whose faces appear on banknotes?

The file `people-on-banknotes.csv` contains data about individuals featured on banknotes from 38 countries. This dataset spans all 22 subregions and sub-subregions of the world, as defined by the United Nations Statistics Division's geoscheme.

It profiles 241 people, detailing their occupations and the year they first appeared on a banknote. Additionally, it includes their year of death — or `NaN` if they were still alive when the dataset was compiled.

Most banknotes were issued after the featured individual’s death. The column `first_death_diff` calculates the difference between the year of their first appearance on a banknote and their year of death (or remains `NaN` if the person was still living at the time of curation).




**NOTE: This project originated from an open-ended project in Cisco Networking Academy's *Data Science Essentials with Python: Coding for Answers* course. I used the provided project as a starting point and independently extended the analysis.**

```python
import pandas as pd

df = pd.read_csv('people-on-banknotes.csv')
df
```

### Quick cleaning

The same person can appear on multiple banknotes. Below we drop the `value` column and remove duplicate people.

```python
df = df.drop(columns=['value'])
df = df.drop_duplicates(subset="name")
df
```

### Project Ideas

- What proportion of individuals featured are male versus female?

- Are writers or politicians more commonly depicted?

- What percentage of featured individuals are musicians?

- What percentage of banknotes were issued before the person’s death?

- Who is the oldest historical figure in the dataset?

- Which countries feature the oldest historical figures on their banknotes?

- What percentage of individuals died at least 100 years before appearing on a banknote?

- Which individuals appeared on a banknote just one year after their death?


**QUESTION 1: What proportion of individuals featured are male versus female?**

```python
gen_prop = df['gender'].value_counts(normalize=True) * 100 #returns the proportion instead of raw count
gen_prop
```

**ANSWER:** Approximately 78% of the featured persons are male, and around 21.99% are female.

**QUESTION 2: Are writers or politicians more commonly depicted?**

```python
occu_prop = (df['occupation'].value_counts(normalize=True) * 100).sort_values(ascending= False)
occu_prop
```

**ANSWER:** Yes, writers are the most commonly depicted, while politicaians ranks third.

**QUESTION 3: What percentage of featured individuals are musicians?**

**ANSWER:** Approximately 4.98% of the featured individuals are musicians.

**QUESTION 4: What percentage of banknotes were issued before the person’s death?**

```python
bef_death = (df['first_death_diff'] < 0).mean() * 100 
bef_death
```

**ANSWER:** Approximately 3.73% of the banknotes were issued before the individual's death.

**QUESTION 5: Who is the oldest historical figure in the dataset?**

```python
oldest = df.sort_values('death', ascending=True).head(1)
oldest
```

**ANSWER:** The oldest historical figure that appeared on a banknote is Hannibal in a Tunisian Dinar currency. He died in 183 BC and first appeared in a banknote in 2013.

**QUESTION 6: Which countries feature the oldest historical figures on their banknotes?
	- Hint: Group by country and aggregate the year of death using the median. Sort the results.**

```python
#convert first the death column from str to int
df['death'] = df['death'].fillna(0).astype(int)

#proceed coding for the median death year by country
country_oldest = df.groupby('country')['death'].median().reset_index().sort_values('death', ascending=True).head(10)
country_oldest
```

**ANSWER:** The top 5 countries that featured the oldest historical figures are South Korea, São Tomé and Príncipe, Ukraine, United States, and Bolivia.

**QUESTION 7: What percentage of individuals died at least 100 years before appearing on a banknote?**

```python
more_hund = (df['first_death_diff'] >= 100).mean() * 100
more_hund
```

**ANSWER:** About 33.20% of the individuals died at least 100 years before they appreared on a banknote. 

**QUEESTION 8: Which individuals appeared on a banknote just one year after their death?**

```python
one_year_death = df.query('first_death_diff == 1')
one_year_death
```

**ANSWER:** Grabriel Garcia Marquez, General Murtula Mohammed, Corazon C. Aquino, and Manuel A. Roxas died just 1 year after they appeared in a banknote.
